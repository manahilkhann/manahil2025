# Legal and Ethical Concerns in Computer Science

## Legal Concerns

### Intellectual Property (IP)
- Creations of the mind: inventions, art, designs, symbols.
- Grants exclusive rights to creators.
- Digitization makes sharing easier but raises issues of ownership and unauthorized use.
- Types of IP protection:
  - **Copyright**: protects original works automatically; registration adds benefits.
  - **Patents**: protect inventions.
  - **Trademarks**: protect brand names/logos.
  - **Trade Secrets**: protect confidential info.

### Licensing a Repository
- A **software license** defines how others can use, modify, and share your code.
- Without a license, default is "All Rights Reserved" — no legal use allowed.
- Licensing enables collaboration, protects work, and prevents legal issues.

### Open Source vs. Open Access
- **Open Source Software (OSS):** code publicly available to use, modify, distribute.
- **Open Access Code:** freely available research with minimal restrictions.
- Advantages:
  - Encourages collaboration and sharing.
- Disadvantages:
  - Can be misused to harm others.

### Common Open-Source Licenses

| License        | Permissions                            | Restrictions                        | Common Uses                 |
|----------------|-------------------------------------|-----------------------------------|-----------------------------|
| MIT            | Use, modify, distribute with credit | No warranty/liability             | Web apps, libraries         |
| Apache 2.0     | Like MIT + patent protection          | License notice required           | Large projects, APIs        |
| GPL            | Use with open-source derivative works| Must keep derivative works open-source | Software meant to stay open |
| BSD 3-Clause   | Free use/modification, no author promotion | No warranty                  | Academic projects           |
| Creative Commons (CC0, CC-BY) | Custom usage rights (not for code) | Mainly docs, art, books         | Documentation, art          |

### Examples of License Use
- Non-profit: GPL for copyleft protection.
- Indie developer: MIT for free sharing with credit.
- Big tech: Apache 2.0 for patent protection and open development.

### Breaking the Law
| Action                                | Reason                                   | License Violated              |
|-------------------------------------|-----------------------------------------|------------------------------|
| Using code without a license         | Default “All Rights Reserved”            | Any repo without license      |
| Modifying GPL code without sharing   | GPL requires derivatives to be open     | GPL                          |
| Removing license notices             | Required by MIT, Apache, GPL              | MIT, Apache 2.0, GPL          |
| Selling open-source as own           | Must follow license terms                 | GPL, Apache 2.0, CC           |
| Using Non-Commercial license for profit | NC licenses prohibit commercial use    | Creative Commons NC           |
| Using proprietary software without license | Requires paid license                   | Proprietary software          |

## Ethical Concerns

- **Respect IP:** Using code without permission disrespects creators.
- **Trust:** Misuse damages the open-source community trust.
- **Fairness:** Using unlicensed code creates unfair advantage.
- **Attribution:** Failing to credit authors is plagiarism.
- **Commercial exploitation:** Profiting without respect to license terms is unethical.

### Real Case: Google vs Oracle
- Oracle claimed Google copied APIs.
- Google argued fair use due to licensing and APIs not being copyrightable.

### Plagiarism and Code Theft
- Wrong:
  - Removing credit and claiming code as your own.
  - Copying from GitHub/Stack Overflow without credit.
  - Selling modified open-source code without permission.
- Right:
  - Fork with credit.
  - Add comments/readme credit.
  - Follow license terms.
