# Graphs Lesson

## What is a Graph?
- A data structure modeling relationships between objects.
- **Nodes (Vertices):** Objects
- **Edges:** Connections between nodes
- **Complete graph:** Every pair of distinct vertices connected by a single edge, no loops or multiple edges.

## Common Uses
- Pathfinding (e.g., GPS, Google Maps)
- Web page ranking (Google PageRank)
- Network routing (data packets)

## Warm-Up Activity
- Delivery driver with 10 packages and houses.
- Questions:
  - How to plan an efficient route?
  - How do real companies solve this?
  - Is trying all routes practical? (Hint: No)

## Representing Graphs

### Adjacency Matrix
- 2D matrix of 0’s and 1’s
- **Undirected graph:** symmetric matrix
- **Directed graph:** matrix with direction-specific entries

### Adjacency List
- Array of lists
- Each index represents a vertex
- Each list contains neighbors (adjacent vertices)
