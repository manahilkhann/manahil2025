# Base 2 Math and Logic Gates

## Binary (Base-2) Number System
- Uses digits **0** and **1**.
- Computers use binary because electronic circuits have two states: OFF (0) and ON (1).
- Each digit is a **bit**.
- Groups of bits represent numbers, text, images, and more.

## Why Binary?
- Easier to store, compress, and transfer data.
- Reliable for electronic processing.
- Used in file formats like JPEG, PNG.

## Number Systems Comparison

| Decimal (Base 10) | Binary (Base 2) | Octal (Base 8) | Hexadecimal (Base 16) |
|-------------------|-----------------|----------------|-----------------------|
| 0                 | 0               | 0              | 0                     |
| 1                 | 1               | 1              | 1                     |
| 2                 | 10              | 2              | 2                     |
| 3                 | 11              | 3              | 3                     |
| 4                 | 100             | 4              | 4                     |
| 8                 | 1000            | 10             | 8                     |
| 15                | 1111            | 17             | F                     |

- **Octal:** Used in early computing.
- **Hexadecimal:** Used in memory addresses, color codes; more compact than binary.

## Reading Binary Numbers
- Each bit represents a power of 2 (from right to left).
- Example: `1010101` → 1×2⁶ + 0×2⁵ + ... + 1×2⁰ = 85 decimal.

## Decimal to Binary Conversion
1. Divide decimal number by 2.
2. Record remainder (0 or 1).
3. Repeat until quotient is 0.
4. Read remainders bottom to top.

Example: 13 → `1101`
