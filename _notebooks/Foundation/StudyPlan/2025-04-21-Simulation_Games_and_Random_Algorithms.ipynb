# Random Number Generation (RNG)

- **Purpose:** Used in simulations, games, security, and testing.
- **Key function:** `random.randint(a, b)` generates random int between `a` and `b` inclusive.
- **Applications:**  
  - Games: dice rolls, card shuffling  
  - Simulations: weather, traffic  
  - Security: passwords, encryption  
  - Statistics: sampling  

---

## Examples

```python
import random

# Random integer 1-10
print(random.randint(1, 10))

# Coin flip
def coin_flip():
    return "Heads" if random.randint(1, 2) == 1 else "Tails"

# Weighted spinner
def color_spinner():
    spin = random.randint(1, 8)
    if spin <= 3:
        return "Green"
    elif spin <= 5:
        return "Blue"
    elif spin == 6:
        return "Red"
    elif spin == 7:
        return "Orange"
    else:
        return "Purple"

# [Popcorn and Homework Hacks](http://127.0.0.1:4100/manahil2025/2025/03/21/lessons_hw_hacks_IPYNB_2_.html)
