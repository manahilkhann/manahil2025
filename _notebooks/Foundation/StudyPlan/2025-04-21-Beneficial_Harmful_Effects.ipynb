# Benefits of Computing Innovations
- **Phone Trees**: Save time, cut costs.
- **Medical Tech**: Aids diagnosis, treatment.
- **Education Access**: Info online = more learning opportunities.

# Harms of Computing Innovations
- **Cyberbullying**: Mental health issues.
- **Social Media**: Anxiety, low self-esteem.
- **Automated Cars**: Ethical risks (trolley problem).

# Neutral Tech (Pros & Cons)
- **AI**: Boosts productivity; harms environment, spreads misinformation.
- **Drones**: Aid rescue; raise privacy concerns.
- **Gene Editing**: Cures disease; ethical worries.
