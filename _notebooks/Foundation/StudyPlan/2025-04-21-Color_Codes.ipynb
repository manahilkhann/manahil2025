## Color Codes and Base64

### 1. Color Codes: The Language of Digital Color

Colors on screens use specific codes that computers understand.

**Hex Codes**  
- Format: `#RRGGBB` or `#RRGGBBAA` (with transparency)  
- Hexadecimal digits: 0–9, A–F  
- Examples:  
  - `#000000` = Black  
  - `#FFFFFF` = White  
  - `#FF5733` = Reddish-orange  

**RGB**  
- Format: `rgb(R, G, B)`  
- Each value: 0 to 255  
- Example: `rgb(255, 87, 51)` = `#FF5733`  

**RGBA**  
- Adds alpha (transparency) channel  
- Format: `rgba(R, G, B, A)` where A = 0 (transparent) to 1 (opaque)  
- Example: `rgba(255, 87, 51, 0.5)` = 50% transparent orange  

| Color | Hex      | RGB           | RGBA               |
|-------|----------|---------------|--------------------|
| Red   | #FF0000  | rgb(255, 0, 0)| rgba(255, 0, 0, 1) |
| Blue  | #0000FF  | rgb(0, 0, 255)| rgba(0, 0, 255, 0.3)|
| White | #FFFFFF  | rgb(255, 255, 255)| rgba(255, 255, 255, 1)|

---

### 2. What Makes a Digital Image?

A digital image is made of pixels, small squares arranged in a grid. Each pixel stores color data in RGB or RGBA format.

Example pixels:  
- Pixel 1: R=255, G=87, B=51 → `#FF5733`  
- Pixel 2: R=0, G=0, B=0 → `#000000`  
- Pixel 3: R=255, G=255, B=255 → `#FFFFFF`  

---

### 3. Types of Image File Formats

| Format | Best Use            | Supports Transparency | Compression Type |
|--------|---------------------|----------------------|------------------|
| JPG    | Photos, web images  | No                   | Lossy            |
| PNG    | Graphics, logos     | Yes                  | Lossless         |
| GIF    | Simple animations   | Yes (limited)        | Lossless (limited)|
| SVG    | Icons, illustrations| Yes                  | Vector           |

**Notes:**  
- JPG compresses by losing some quality, good for photos.  
- PNG keeps full quality, good for logos and transparency.  
- GIF supports simple animations with limited colors.  
- SVG is vector-based and scales without quality loss.

---

### 4. How Images Store Color

Each pixel stores color as three components:  
- Red (R): 0–255  
- Green (G): 0–255  
- Blue (B): 0–255  
- Sometimes Alpha (A): 0.0–1.0 for transparency  

Example:  
- R=76, G=175, B=80  
- Hex: `#4CAF50`  
- RGBA: `rgba(76, 175, 80, 1)`

---

### 5. Base64: Turning Binary into Text

**What is Base64?**  
A way to encode binary data (images, files) into text using 64 characters: A–Z, a–z, 0–9, +, and /. `=` is used for padding.

**Why use it?**  
Some systems (emails, URLs, APIs) can only handle text, so Base64 safely transmits binary data.

**How it works (conceptually):**  
- Converts binary into 6-bit groups.  
- Maps each 6-bit group to one Base64 character.
