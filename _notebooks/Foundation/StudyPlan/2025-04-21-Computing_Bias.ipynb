# Big Idea 5: Computing Bias – AP Exam Review

## What is Computing Bias?
- **Bias**: When algorithms unfairly favor or disadvantage certain groups.
- Causes: Biased data, flawed design, or human error.

---

## Example: Netflix Bias
- **Majority Preference Bias**: Recommends only popular content.
- **Filtering Bias**: Limits recommendations based on past behavior.

---

## Types of Data
- **Explicit Data**: Directly provided by users (e.g., preferences).
- **Implicit Data**: Inferred from actions (e.g., watch history).

---

## Types of Bias
- **Algorithmic**: Faulty systems (e.g., biased hiring algorithms).
- **Data Bias**: Skewed or incomplete training data.
- **Cognitive Bias**: Human bias in data collection.

---

## Intentional vs. Unintentional Bias
- **Intentional**: Deliberate unfairness (e.g., biased algorithm design).
- **Unintentional**: Happens due to flawed data (e.g., recognition software issues).

---

## Mitigation Strategies
1. **Pre-processing**: Clean and diversify data.
2. **In-processing**: Adjust algorithms during training.
3. **Post-processing**: Monitor and adjust model output.
