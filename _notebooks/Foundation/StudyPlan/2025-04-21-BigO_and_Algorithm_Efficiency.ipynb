# Big O & Algorithm Efficiency

## Why Efficiency Matters
- **Fast Apps**: Less lag, smoother use.
- **Resource Saving**: Uses less memory, CPU, energy.
- **Scalability**: Works better on big data.

## Big O Notation
- **O(1)** – Constant time (e.g., array lookup)
- **O(log n)** – Logarithmic (e.g., binary search)
- **O(n)** – Linear (e.g., loop through list)
- **O(n log n)** – Efficient sorting (e.g., merge sort)
- **O(n²)** – Slower on big data (e.g., bubble sort)

## Example: Even/Odd Check
- Best methods:  
  - `num % 2 == 0`  
  - Check last digit with `in "02468"`

## Search Algorithms
- **Linear Search (O(n))**: Checks each item  
- **Binary Search (O(log n))**: Halves list each step (needs sorting)

## Sort Algorithms
- **Bubble Sort (O(n²))**: Simple, but slow  
- **Merge Sort (O(n log n))**: Fast on big lists

## Key Trade-Offs
- Faster ≠ Less Memory  
- Choose based on your use case (speed vs. space)

## Final Tips
- Big O helps you compare code efficiency  
- Try writing and testing these algorithms yourself!
