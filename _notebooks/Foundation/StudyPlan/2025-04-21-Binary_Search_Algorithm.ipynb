# What Is Binary?

- Binary is a base-2 number system made of 0s and 1s.  
- Computers use binary to store data (text, images, video, etc.).  
- 1 byte = 8 bits → can store 256 values.

# Real-World Uses

- RGB pixels in images  
- ASCII and Unicode text encoding  
- Streaming music and video

# Converting Binary & Decimal

**Decimal to Binary**  
- Divide by 2 repeatedly  
- Record remainders  
- Read remainders bottom-up  

**Binary to Decimal**  
- Multiply each binary digit by powers of 2  
- Add the values

# Binary vs Linear Search

- **Binary Search**: Efficient method for finding items in sorted lists  
  - Repeatedly divides the list in half  
  - Time complexity: `O(log n)`  

- **Linear Search**: Checks each item one by one  
  - Slower  
  - Time complexity: `O(n)`

# Comparison Table

| Feature            | Binary Search     | Linear Search       |
|--------------------|-------------------|---------------------|
| Requires sorted?   |  Yes              |  No               |
| Speed (100 items)  | ~7 steps          | Up to 100 steps     |
| Best for…          | Large sorted data | Small/unsorted data |

# Real-Life Analogy

- Like finding a word in a dictionary by flipping to the middle, not page by page.
