# List Creation

```js
def get(self):            
            posts = firstPlaceLeaderboard.query.all()
            json_ready = [post.read() for post in posts]
            return jsonify(json_ready)
```

Explanation:
This python function gathers all leaderboard posts from the firstPlaceLeaderboard database table using query.all(), and processes each one by calling its .read() method. This turns each database object into a dictionary, creating a full list of readable leaderboard entries.

---

# List Process

```js
json_ready = [post.read() for post in posts]
```

Explanation:
This line creates a new list called json_ready by iterating over each post in the posts list (a for loop inside [] in python is called "list comprehension"). Each post is transformed via .read() method into a dictionary, making the data JSON serializable.
