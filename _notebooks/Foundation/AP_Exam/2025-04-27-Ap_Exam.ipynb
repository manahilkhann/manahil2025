```python
plant_type = input("What type of plant do you have?")
if plant_type == "cactus":
    print("Cacti need very little water. Water them once every 2-3 weeks.")
elif plant_type == "fern":
    print("Ferns like to be kept moist. Water them once a week.")
elif plant_type == "bamboo":
    print("Bamboo needs a lot of water. Water them every 2-3 days.")
elif plant_type == "succulent":
    print("Succulents need moderate water. Water them every 1-2 weeks.")
elif plant_type == "orchid":
    print("Orchids like to dry out between waterings. Water them every 1-2 weeks.")
else:
    print("Sorry, I don't have information on that type of plant.")
# It uses if-elif statements to check the plant type and print the corresponding watering instructions.
# The code also includes a default case to handle unknown plant types.
# The watering instructions are based on common care guidelines for each type of plant.
# The code can also be enhanced by providing additional information such as light requirements and temperature preferences for each plant type.
# The code can be made more interactive by allowing users to input multiple plant types and get care instructions for each.
```
