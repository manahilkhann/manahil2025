# The Panelists

## David Tomasz
FBI Intelligence Analyst in San Diego’s cyber squad, David specializes in ransomware and international cyber threats, including foreign IP theft. He’s a key member of the San Diego Cyber Council and emphasizes the value of networking, leadership, and adapting to the evolving tech work landscape.

## Tricia Mercer
Chief People Officer at INDUS and board member of the Cyber Center of Excellence, Tricia brings 20+ years in cyber defense. She focuses on talent development, workplace culture, and bridging technical and business communication. She's passionate about cybersecurity fundamentals and phishing defense.

## Victor Nzeata
Founder of Cyber Brain Academy and the Army Reserve’s first Cyber Operations Officer, Victor trains transitioning service members in cybersecurity. With expertise in AI security and insider threat prevention, he advocates for specialization, formal education, and awareness of zero-day vulnerabilities.

# Key Takeaways

- Be someone people like, this will allow you do deepen you connections and increase opportunities
- You have to be curious and passionate about your job to do well 
- Finding a mentor and getting internships will help you drasticly when pursuing your career
- Networking is very important to furthur your opportunities
- Be a people person
- You cant go home after work and think "im done", the best people are those who go home, put their kids to sleep then keep working and learning
- You will always have bad people in your group, you have to learn to manage them and get your own work done all together
- You must have integrity and honesty
- Success in cybersecurity relies on a mix of strong technical skills and strong people skills.

# My Reflection

This panel gave me a new point of view of this workforce and a deeper understanding of what it truly means to work in cybersecurity. It taught me that in this field it is important to keep a balance between technical knowledge as well as people skills. I had previously never really thought about the importance of connections or being a people person but I learnt that it is a crucial aspect of widening your opportunities. Additionally, not only is cybersecurity about securing systems, it has a handful of other aspects such as critical thinking, being ethical, adapting quickly, and staying committed to making a meaningful difference.
